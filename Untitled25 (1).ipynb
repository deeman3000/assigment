```python
import os
import yaml
from ultralytics import YOLO
from google.colab import files # For downloading the file

# 1. SETUP PATHS
def get_yolo_path(folder_name):
    for root, dirs, files in os.walk("/content/traffic_data"):
        if folder_name in root and any(f.endswith(('.jpg', '.png', '.jpeg')) for f in files):
            return root
    return None

train_images = get_yolo_path("train")
valid_images = get_yolo_path("valid") or get_yolo_path("test")

if not train_images or not valid_images:
    print(" ERROR: Could not find image folders.")
else:
    # 2. CREATE THE YAML
    data_yaml = {
        'train': train_images,
        'val': valid_images,
        'nc': 4,
        'names': ['Helmet', 'No-Helmet', 'Triple-Riding', 'Other-Violation']
    }

    yaml_path = '/content/traffic_data/data.yaml'
    with open(yaml_path, 'w') as f:
        yaml.dump(data_yaml, f)

    # 3. RUN TRAINING
    model = YOLO('yolov8s.pt')

    # We save the results to a variable so we can find the save path easily
    results = model.train(
        data=yaml_path,
        epochs=20,
        imgsz=640,
        batch=16,
        name='traffic_final_model',
        device=0
    )

    # 4. EXPORT / FIND THE .PT FILE
    # YOLO saves to: runs/detect/traffic_final_model/weights/best.pt
    best_model_path = f'{results.save_dir}/weights/best.pt'

    print(f"\n TRAINING FINISHED!")
    print(f" Your weights are saved at: {best_model_path}")

    # Optional: Copy to your main content folder for easy access
    !cp {best_model_path} /content/my_traffic_model.pt
    print(f" A copy has been created at: /content/my_traffic_model.pt")

    # Optional: Uncomment the line below if you want Colab to trigger a
    # browser download of the model file immediately when finished.
    # files.download('/content/my_traffic_model.pt')
```
